```python
import pandas as pd
import csv
import os
```

```python
class student:
    def __init__ (self,Name, Roll, Class_Name, Address):
        self.Name=Name
        self.Roll=Roll
        self.Class_Name=Class_Name
        self.Address=Address
```

```python
class student_database:
    def load_data (self):
        if not os.path.exists("student_auth.csv"):
            return None
        return pd.read_csv("student_auth.csv")
    
    def check_student(self,student):
        read=self.load_data()
        if read is None:
            return None
        for index,row in read.iterrows():
            if row["Name"]==student.Name and row["Roll"]==student.Roll:
                return row,index
            
        return None
        

    def add_student(self,student):
        read=self.load_data()
        if read is None:
            read=pd.DataFrame(columns=["Name","Roll","Class_Name","Address"])
            read.to_csv("student_auth.csv",index=False)
        if self.check_student(student):
            print("Student is already in database")  
            return
        with open ("student_auth.csv","a",newline="") as data:
            writer=csv.writer(data)
            writer.writerow([student.Name,student.Roll,student.Class_Name,student.Address])
            print("Student is successfully added")
            return
            

    def remove_student(self,student):
        read=self.load_data()
        if read is None:
            print("There is no database")
            return
        if not self.check_student(student):
            print("Student is not in the database")
            return

        _,index=self.check_student(student)
        read=read.drop(index)
        read.to_csv("student_auth.csv",index=False)
        print("successfully remove")
        return


    def get_info(self,student):
        read=self.load_data()
        if read is None:
            print("There is no database")
            return
        if not self.check_student(student):
            print("Student is not in database")
            return
        row,_=self.check_student(student)
        print(row)
        return

    def all_student(self):
        read=self.load_data()
        if read is None:
            print("There is no database")
        print(read)
    def update_info(self,student,new_student):
        read=self.load_data()
        if read is None:
            print("There is no database")
            return
        student_info=self.check_student(student)
        if student_info is None:
            print("Studnet is not in the database")
            return
        _,index=student_info
        read.loc[index,"Name"]=new_student.Name
        read.loc[index,"Roll"]=new_student.Roll
        read.loc[index,"Class_Name"]=new_student.Class_Name
        read.loc[index,"Address"]=new_student.Address
        read.to_csv("student_auth.csv",index=False)
        print("student information is successfully updated")
        return
    

        
```

```python
class account_database:
    def file_check(self):
        if not os.path.exists("account.csv"):
            return None
        return pd.read_csv("account.csv")
    
    def check_valid(self,account):
        read=self.file_check()
        if read is None:
            return None
        for index,row in read.iterrows():
            if row["username"]==account.username and row["password"]==account.password and row["role"]==account.role:
                return index,row
        return None


    def create_data(self,account):
        read=self.file_check()
        if read is None:
            read=pd.DataFrame(columns=["username","password","role"])
            read.to_csv("account.csv",index=False)
        with open ("account.csv","a",newline="") as data:
            writer=csv.writer(data)
            writer.writerow([account.username,account.password,account.role])
            print("Successfully Added")
            return

    def login(self,account):
        read=self.file_check()
        if read is None:
            print("There is no database")
            return
        if self.check_valid(account):
            print("succcessfull login")
            return
        print("Invalid username/password")
        return
        
    def register(self,account):
        read=self.file_check()
        if read is None:
            print("There is no database")
            return
        if not self.check_valid(account) is None:
            print("user already registed")
            return
        self.create_data(account)
        return
        

    def change_password(self,account,new_password):
        read=self.file_check()
        if read is None:
            print("There is no database")
            return 
        valid=self.check_valid(account)
        if valid is None:
            print("user is not valid")
            return
        index,_=valid
        read.loc[index,"password"]=new_password.password
        read.to_csv("account.csv",index=False)
        print("Password is successfully changed")
        return
        
    def delete_account(self,account):
        read=self.file_check()
        if read is None:
            print("There is no database")
            return
        valid=self.check_valid(account)
        index,_=valid
        read.drop(index)
        read.to_csv("account.csv",index=False)
        print("successfully deleted")
        return

        
```
