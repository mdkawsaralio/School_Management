```python
import pandas as pd
import csv
import os
```

```python
class student:
    def __init__ (self,Name, Roll, Class_Name, Address):
        self.Name=Name
        self.Roll=Roll
        self.Class_Name=Class_Name
        self.Address=Address
```

```python
class database:
    def load_data (self):
        if not os.path.exists("student_auth.csv"):
            return None
        return pd.read_csv("student_auth.csv")
    
    def check_student(self,student):
        read=self.load_data()
        if read is None:
            return None
        for index,row in read.iterrows():
            if row["Name"]==student.Name and row["Roll"]==student.Roll:
                return row,index
            
        return None
        

    def add_student(self,student):
        read=self.load_data()
        if read is None:
            read=pd.DataFrame(columns=["Name","Roll","Class_Name","Address"])
            read.to_csv("student_auth.csv")
        if self.check_student(student):
            print("Student is already in database")  
            return
        with open ("student_auth.csv","a",newline="") as data:
            writer=csv.writer(data)
            writer.writerow([student.Name,student.Roll,student.Class_Name,student.Address])
            print("Student is successfully added")
            return
            

    def remove_student(self,student):
        read=self.load_data()
        if read is None:
            print("There is no database")
            return
        if not self.check_student(student):
            print("Student is not in the database")
            return

        _,index=self.check_student(student)
        read=read.drop(index)
        read.to_csv("student_auth.csv",index=False)
        print("successfully remove")
        return


    def get_info(self,student):
        read=self.load_data()
        if read is None:
            print("There is no database")
            return
        if not self.check_student(student):
            print("Student is not in database")
            return
        row,_=self.check_student(student)
        print(row)
        return

    def all_student(self):
        read=self.load_data()
        if read is None:
            print("There is no database")
        print(read)
        
```
